```python
import pandas as pd
import numpy as np
import pyxlsb 
```

```python
stctv = pd.read_excel("stc TV Data Set_T1.xlsb")
```

```python
stctv.head()
```

```python
stctv.drop(columns="Column1", inplace=True)
stctv.shape
```

```python
stctv.info()
```

```python
stctv["date_"] = pd.to_datetime(stctv["date_"], unit= "D", origin="30/12/1899")
```

```python
stctv.head()
```

```python
stctv["user_id_maped"] = stctv["user_id_maped"].astype(str)
```

```python
stctv.info()
```

```python
stctv.isnull().any()
```

```python
stctv["program_name"] = stctv["program_name"].str.strip()
stctv["program_desc"] = stctv["program_desc"].fillna("UNKNOWN")
stctv.isnull().any()
```

```python
stctv.describe()
```

```python
detailedstctv = stctv.groupby(["program_name","program_class","season","episode"]).agg(
    no_of_users_watched = ("user_id_maped", "nunique") , 
    no_of_times_watched = ("user_id_maped", "count") , 
    total_watch_hours = ("duration_seconds", "sum")
)
detailedstctv["total_watch_hours"] = detailedstctv["total_watch_hours"] / 3600
detailedstctv = detailedstctv.reset_index()
detailedstctv["program_name"] = np.where(detailedstctv["program_class"] == "SERIES/EPISODES", detailedstctv["program_name"] + " S" + (detailedstctv["season"]).astype(str) +" EP" + (detailedstctv["episode"]).astype(str),detailedstctv["program_name"])
detailedstctv.drop(columns=["season", "episode"], inplace=True)
detailedstctv
```

```python
detailedstctv.describe()
```

```python
moredetailedstctv = stctv.groupby(["program_class"]).agg(
    no_of_users_watched = ("user_id_maped", "nunique") , 
    no_of_times_watched = ("user_id_maped", "count") , 
    total_watch_hours = ("duration_seconds", "sum")
).reset_index()
moredetailedstctv["total_watch_hours"] = moredetailedstctv["total_watch_hours"]/3600
moredetailedstctv
```

```python
import matplotlib.pyplot as plt
import plotly 
import plotly.express as px
import plotly.graph_objects as go
from plotly.subplots import make_subplots
```

```python
figureOne = px.pie(moredetailedstctv, names= "program_class", values= "no_of_users_watched", title="Users Watched: Movies vs Series")
figureOne.update_traces(textinfo= "percent+value")
figureOne.show()
figureTwo = px.pie(moredetailedstctv, names="program_class", values="total_watch_hours",color_discrete_sequence=["Teal","Pink"], title="Total Hours Watched: Movies vs Series")
figureTwo.update_traces(textinfo = "value")
figureTwo.show()
```

```python
figureThree = px.bar(moredetailedstctv,x ="program_class", y = "no_of_times_watched", color_discrete_sequence=["#234061"], width=500)
figureThree.update_layout(bargap=0.7)
figureThree.show()
```

```python
detailedstctv.sort_values(by=["no_of_users_watched", "no_of_times_watched",	"total_watch_hours"], ascending=False, inplace=True)
detailedstctv.reset_index(drop=True, inplace=True)
figureFour = px.pie(detailedstctv.head(10), names="program_name", values="total_watch_hours", title="Top 10 Movie/Series")
figureFour.show()
```

```python
resolutionSummary = stctv.groupby(["program_class","hd"]).agg(
    no_of_users_watched = ("user_id_maped", "nunique") , 
    no_of_times_watched = ("user_id_maped", "count") , 
    total_watch_hours = ("duration_seconds", "sum")
).reset_index()
resolutionSummary["total_watch_hours"] = resolutionSummary["total_watch_hours"] / 3600
resolutionSummary["hd"] = np.where(resolutionSummary["hd"] == 1, "HD", "SD")
resolutionSummary
```

```python
color_map = {"MOVIE": "#631452", "SERIES/EPISODES": "#E67E22"}
figureFive = px.bar(resolutionSummary, x ="hd", y ="no_of_times_watched", color="program_class", barmode="stack", title="Total Watches: HD vs SD by Program Class", color_discrete_map= color_map)
figureSix = px.bar(resolutionSummary, x ="hd", y ="no_of_users_watched", color="program_class", barmode="stack", title="Unique Viewers: HD vs SD by Program Class", color_discrete_map= color_map)
figureSeven = px.bar(resolutionSummary, x="hd", y= "total_watch_hours", color = "program_class", barmode="group", title="Total Watch Hours: HD vs SD by Program Class", color_discrete_map= color_map)
figureFive.show()
figureSix.show()
figureSeven.show()
```
