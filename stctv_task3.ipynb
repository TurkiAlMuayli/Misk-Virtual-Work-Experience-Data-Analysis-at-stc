```python
import numpy as np
import pandas as pd
import pyxlsb
```

```python
stctvR = pd.read_excel("stc TV Data Set_T3.xlsx")
stctvOG = pd.read_excel("stc TV Data Set_T1.xlsb")
```

```python
stctvR.drop("Unnamed: 0", axis=1, inplace=True)
```

```python
stctvOG.head()
```

```python
stctvR.isnull().any()
```

```python
stctvOG.drop(["Column1","date_","season", "episode", "series_title","hd","original_name"], axis=1, inplace=True)
```

```python
stctvOG["program_name"] = stctvOG["program_name"].str.strip()
```

```python
userProgramRatingMatrix = stctvR.pivot_table(index = "user_id_maped", columns="program_name", values="rating")
userProgramRatingMatrix
```

```python
prog_median_duration = stctvOG.groupby("program_name")["duration_seconds"].median()
```

```python
stctvOG["watching_rating"] = stctvOG["duration_seconds"] / stctvOG["program_name"].map(prog_median_duration)
stctvOG["watching_rating"].clip(upper=4, inplace=True)
```

```python
stctvOG["watching_rating"] = stctvOG["watching_rating"].round().astype(int)
```

```python
userProgramDurationMatrix = stctvOG.pivot_table(index="user_id_maped", columns="program_name", values="watching_rating")
userProgramDurationMatrix
```

```python
userProgramRatingMatrix.fillna(0, inplace=True)
userProgramDurationMatrix.fillna(0, inplace=True)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_ratings = cosine_similarity(userProgramRatingMatrix.T)
similarity_duration = cosine_similarity(userProgramDurationMatrix.T)
```

```python
similarity_ratings_df = pd.DataFrame(similarity_ratings, index=userProgramRatingMatrix.columns, columns=userProgramRatingMatrix.columns)
similarity_ratings_df
```

```python
similarity_duration_df = pd.DataFrame(similarity_duration, index=userProgramDurationMatrix.columns , columns=userProgramDurationMatrix.columns)
similarity_duration_df
```
